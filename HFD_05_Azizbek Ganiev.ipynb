## Exercise 5.1

```
import pandas as pd

# Data loaded from previous lab work
df = pd.read_parquet("dataUSA_AAPL_META_after_lab4.parquet")

df['SMA'] = df['close'].rolling(20).mean()
df['signal'] = 0
df.loc[df['close'] > df['SMA'], 'signal'] = 1
df.loc[df['close'] < df['SMA'], 'signal'] = -1
df.head()
```

## Exercise 5.2

```
df['ret'] = df['close'].pct_change()
df['position'] = df['signal'].shift(1)
df['pnl'] = df['position'] * df['ret']
df['cum_pnl'] = df['pnl'].cumsum()
df[['position','pnl','cum_pnl']].head()
```

## Exercise 5.3

```
df['EMA'] = df['close'].ewm(span=15).mean()
df['signal2'] = 0
df.loc[df['close'] > df['EMA'], 'signal2'] = 1
df.loc[df['close'] < df['EMA'], 'signal2'] = -1

df['position2'] = df['signal2'].shift(1)
df['pnl2'] = df['position2'] * df['ret']
df['cum_pnl2'] = df['pnl2'].cumsum()
df[['position2','pnl2','cum_pnl2']].head()
```

## Exercise 5.4 (Double Volatility Breakout)

```
df['vol'] = df['close'].pct_change().rolling(20).std()

up = df['close'].shift(1) + 2*df['vol']
down = df['close'].shift(1) - 2*df['vol']

df['signal3'] = 0
df.loc[df['close'] > up, 'signal3'] = 1
df.loc[df['close'] < down, 'signal3'] = -1

df['position3'] = df['signal3'].shift(1)
df['pnl3'] = df['position3'] * df['ret']
df['cum_pnl3'] = df['pnl3'].cumsum()

df[['position3','pnl3','cum_pnl3']].head()
```
