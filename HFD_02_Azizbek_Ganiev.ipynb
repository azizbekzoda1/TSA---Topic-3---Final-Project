```
# HFD_02 Homework
# Name: Azizbek Ganiev
# Student ID: 475150
# Course: Quantitative Strategies and High-Frequency Data

# --- Import libraries ---
import pandas as pd
import matplotlib.pyplot as plt

# --- Load the data ---
data = pd.read_pickle('dataUSA_AAPL_META.pkl')
data.head()

# --- Exercise 2.1 ---
# Create a dataset with closing prices and returns, and plot the graphs.

# Select closing prices for AAPL and META
prices = data[['AAPL', 'META']].copy()

# Calculate simple returns
returns = prices.pct_change().dropna()

# Plot closing prices
plt.figure(figsize=(8,4))
plt.plot(prices['AAPL'], label='AAPL')
plt.plot(prices['META'], label='META')
plt.title('Closing Prices of AAPL and META')
plt.legend()
plt.show()

# Plot returns
plt.figure(figsize=(8,4))
plt.plot(returns['AAPL'], label='AAPL Returns')
plt.plot(returns['META'], label='META Returns')
plt.title('Rates of Return')
plt.legend()
plt.show()

# --- Exercise 2.2 ---
# Calculate and plot rolling std of (r1 - r2) with 150-min window.

returns_diff = returns['AAPL'] - returns['META']
roll_std = returns_diff.rolling(window=150).std()

plt.figure(figsize=(8,4))
plt.plot(roll_std, color='purple')
plt.title('Rolling Std (150-min) of (AAPL - META) Returns')
plt.show()

# --- Exercise 2.3 ---
# Plot moving correlation (90-min window)

rolling_corr = returns['AAPL'].rolling(window=90).corr(returns['META'])

plt.figure(figsize=(8,4))
plt.plot(rolling_corr, color='green')
plt.title('Rolling Correlation (90-min) between AAPL and META')
plt.show()

# --- Exercise 2.4 ---
# Save dataset for further analysis.

final_data = pd.concat([prices, returns], axis=1)
final_data.columns = ['AAPL_Price', 'META_Price', 'AAPL_Return', 'META_Return']
final_data.to_csv('HFD_02_final_data.csv', index=False)
print('Dataset saved as HFD_02_final_data.csv')
```
