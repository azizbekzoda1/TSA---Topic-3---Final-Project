## Load Data

```

import pandas as pd
df = pd.read_csv("lab6_synthetic_data.csv", parse_dates=["time"])
df = df.set_index("time")
df.head()
```

## Exercise 6 – Basic Preview

```

df[['AAPL','META']].plot(figsize=(10,4), title='Synthetic Prices')
```

## Exercise 6.1 – Single Moving Average Strategy

```

import numpy as np

df['SMA20'] = df['AAPL'].rolling(20).mean()
df['signal'] = np.where(df['AAPL'] > df['SMA20'], 1, -1)
df['ret'] = df['AAPL'].pct_change()
df['pnl'] = df['signal'].shift(1) * df['ret']

sr_gross = df['pnl'].mean()/df['pnl'].std()*np.sqrt(252)
sr_net = sr_gross - 0.5

df[['AAPL','SMA20']].plot(figsize=(10,4),title='SMA20 Strategy')
df['pnl'].cumsum().plot(figsize=(10,4),title='Cumulative PnL')

sr_gross, sr_net
```

## Exercise 6.2 – Two Moving Medians Strategy

```

df['med_fast']=df['AAPL'].rolling(10).median()
df['med_slow']=df['AAPL'].rolling(30).median()
df['signal2']=np.where(df['med_fast']>df['med_slow'],1,-1)
df['pnl2']=df['signal2'].shift(1)*df['ret']

sr2_gross=df['pnl2'].mean()/df['pnl2'].std()*np.sqrt(252)
sr2_net=sr2_gross-0.5

df[['AAPL','med_fast','med_slow']].plot(figsize=(10,4),title='Two Median Strategy')
df['pnl2'].cumsum().plot(figsize=(10,4),title='Cumulative PnL - Two Medians')

sr2_gross, sr2_net
```

## Exercise 6.3 – Volatility Breakout Strategy

```

import numpy as np

df['vol']=df['AAPL'].pct_change().rolling(30).std()
df['upper']=df['vol']*2
df['lower']=-df['vol']*2
df['signal_vb']=df['AAPL'].pct_change()

df['pos_vb']=np.where(df['signal_vb']>df['upper'],1,
               np.where(df['signal_vb']<df['lower'],-1,0))
df['pnl_vb']=df['pos_vb'].shift(1)*df['ret']

sr_vb_gross=df['pnl_vb'].mean()/df['pnl_vb'].std()*np.sqrt(252)
sr_vb_net=sr_vb_gross-0.5

df[['upper','lower','signal_vb']].plot(figsize=(10,4),title='VB Signal')
df['pnl_vb'].cumsum().plot(figsize=(10,4),title='Cumulative PnL - VB')

sr_vb_gross, sr_vb_net
```
